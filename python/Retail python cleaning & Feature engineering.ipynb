```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv(r"C:\Users\hp\Downloads\Sample - Superstore.csv", encoding='latin1')
df
```

```python
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_')
```

```python
print("Shape:", df.shape)
print("\nDtypes:\n", df.dtypes)
print("\nNull values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
print("\nSample:\n", df.head(3))
```

```python
# ── 3. FIX DATA TYPES ────────────────────────────────────────────────────────
df['Order_Date']  = pd.to_datetime(df['Order_Date'], format='mixed')
df['Ship_Date']   = pd.to_datetime(df['Ship_Date'],  format='mixed')
df['Postal_Code'] = df['Postal_Code'].astype(str).str.strip()
```

```python
#Remove duplicate rows
df[df.duplicated(subset = ['Order_ID', 'Product_ID'])]
```

```python
df= df.drop_duplicates(subset=['Order_ID', 'Product_ID'])
```

```python
# ── 4. PRE-CLEAN VALIDATION ──────────────────────────────────────────────────
print("\n--- Pre-Clean Validation ---")
print("Negative/Zero Sales:  ", (df['Sales'] <= 0).sum())
print("Invalid Discounts:    ", ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
print("Ship before Order:    ", (df['Ship_Date'] < df['Order_Date']).sum())

```

```python

# ── 5. CLEAN ─────────────────────────────────────────────────────────────────
# Standardize text columns
text_cols = ['Customer_Name', 'Segment', 'City', 'State', 'Region',
             'Category', 'Sub_Category', 'Product_Name', 'Ship_Mode']
for col in text_cols:
    df[col] = df[col].str.strip().str.title()

# Remove invalid discounts
df = df[(df['Discount'] >= 0) & (df['Discount'] <= 1)].reset_index(drop=True)

# Remove zero or negative sales
df = df[df['Sales'] > 0].reset_index(drop=True)

# Drop duplicates
df.drop_duplicates(inplace=True)

# Drop rows where critical fields are null
df.dropna(subset=['Order_ID', 'Customer_ID', 'Product_ID', 'Sales'], inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

print("\n--- After Cleaning ---")
print("Shape:", df.shape)
print("Nulls:\n", df.isnull().sum())

```

```python
# ── 6. DERIVED COLUMNS ───────────────────────────────────────────────────────
# 1. Profit Margin %
df['Profit_Margin_Pct'] = np.where(
    df['Sales'] > 0,
    (df['Profit'] / df['Sales'] * 100).round(2),
    0
)

# 2. Delivery Days
df['Delivery_Days'] = (df['Ship_Date'] - df['Order_Date']).dt.days

# 3. Discount Band
df['Discount_Band'] = pd.cut(
    df['Discount'],
    bins=[-0.001, 0.001, 0.2, 0.4, 1.0],
    labels=['No Discount', 'Low (1-20%)', 'Medium (21-40%)', 'High (41%+)']
)
df['Discount_Band'] = df['Discount_Band'].astype(str)

# 4. Order Year and Month
df['Order_Year']       = df['Order_Date'].dt.year
df['Order_Month']      = df['Order_Date'].dt.month
df['Order_Month_Name'] = df['Order_Date'].dt.strftime('%b')

# 5. Season
df['Season'] = df['Order_Month'].map({
    12: 'Winter', 1: 'Winter',  2: 'Winter',
    3:  'Spring', 4: 'Spring',  5: 'Spring',
    6:  'Summer', 7: 'Summer',  8: 'Summer',
    9:  'Fall',   10: 'Fall',   11: 'Fall'
})

# 6. Is Loss flag
df['Is_Loss'] = df['Profit'].apply(lambda x: 'Loss' if x < 0 else 'Profit')

# 7. Margin Band
def margin_band(m):
    if m >= 20:   return 'High (>=20%)'
    elif m >= 10: return 'Medium (10-20%)'
    elif m >= 0:  return 'Low (0-10%)'
    else:         return 'Loss'

df['Margin_Band'] = df['Profit_Margin_Pct'].apply(margin_band)

# 8. Sales Bucket
df['Sales_Bucket'] = pd.cut(
    df['Sales'],
    bins=[0, 50, 200, 500, 1000, 25000],
    labels=['Micro (<$50)', 'Small ($50-200)', 'Medium ($200-500)',
            'Large ($500-1K)', 'XL (>$1K)']
)
df['Sales_Bucket'] = df['Sales_Bucket'].astype(str)

# 9. Revenue per Unit
df['Revenue_per_Unit'] = (df['Sales'] / df['Quantity']).round(2)

# 10. Profit per Unit
df['Profit_per_Unit'] = (df['Profit'] / df['Quantity']).round(2)

# 11. Discount Amount
df['Discount_Amount'] = (df['Sales'] * df['Discount']).round(2)

# 12 Ship Speed 
df['Ship_Speed'] = df['Ship_Mode'].map({
    'Same Day': 'Express',
    'First Class': 'Fast',
    'Second Class': 'Standard',
    'Standard Class': 'Economy'
})

# Handle any unexpected values
df['Ship_Speed'] = df['Ship_Speed'].fillna('Economy')

print("\n--- Derived Columns Added ---")
print(df[['Sales', 'Profit', 'Profit_Margin_Pct', 'Delivery_Days',
          'Discount_Band', 'Order_Year', 'Order_Month', 'Order_Month_Name',
          'Season', 'Is_Loss', 'Margin_Band', 'Sales_Bucket',
          'Revenue_per_Unit', 'Profit_per_Unit', 'Discount_Amount','Ship_Speed']].head(5).to_string())
```

```python
# ── 7. PRE-SPLIT VALIDATION ──────────────────────────────────────────────────
conflicts_city    = df.groupby('Customer_ID')['City'].nunique()
conflicts_segment = df.groupby('Customer_ID')['Segment'].nunique()
print("\nCustomers with multiple Cities:   ", (conflicts_city > 1).sum())
print("Customers with multiple Segments: ", (conflicts_segment > 1).sum())
```

```python

# ── 8. SPLIT INTO 3 TABLES ───────────────────────────────────────────────────
# Fact Table
transaction_data = df[[
    'Order_ID', 'Order_Date', 'Ship_Date', 'Ship_Mode', 'Ship_Speed',
    'Customer_ID', 'Product_ID',
    'Sales', 'Quantity', 'Discount', 'Profit',
    'Profit_Margin_Pct', 'Delivery_Days', 'Discount_Band',
    'Order_Year', 'Order_Month', 'Order_Month_Name', 'Season', 'Is_Loss',
    'Margin_Band', 'Sales_Bucket', 'Revenue_per_Unit',
    'Profit_per_Unit', 'Discount_Amount'
]].copy()
print(df[['Ship_Mode', 'Ship_Speed']].drop_duplicates().to_string())

# Dimension — customer
customer_data = df[[
    'Customer_ID', 'Customer_Name', 'Segment',
    'City', 'State', 'Postal_Code', 'Region'
]].drop_duplicates(subset='Customer_ID').reset_index(drop=True)

# Dimension — product
product_data = df[[
    'Product_ID', 'Category', 'Sub_Category', 'Product_Name'
]].drop_duplicates(subset='Product_ID').reset_index(drop=True)


```

```python

# ── 9. POST-SPLIT VALIDATION ─────────────────────────────────────────────────
print("\n--- Table Shapes ---")
print("transaction_data:", transaction_data.shape)
print("customer_data:   ", customer_data.shape)
print("product_data:    ", product_data.shape)

unmatched_cust = transaction_data[~transaction_data['Customer_ID'].isin(customer_data['Customer_ID'])]
unmatched_prod = transaction_data[~transaction_data['Product_ID'].isin(product_data['Product_ID'])]
print("\nUnmatched Customer_IDs:", len(unmatched_cust))
print("Unmatched Product_IDs: ", len(unmatched_prod))

print("\nDelivery Days — negative values:", (transaction_data['Delivery_Days'] < 0).sum())
print("Discount_Band — 'nan' values:   ", (transaction_data['Discount_Band'] == 'nan').sum())
print("Sales_Bucket  — 'nan' values:   ", (transaction_data['Sales_Bucket']  == 'nan').sum())

```

```python
# ── 10. EXPORT ───────────────────────────────────────────────────────────────
output_path = r"C:\Users\hp\Desktop\project\\"

transaction_data.to_csv(output_path + "transaction_data.csv", index=False)
customer_data.to_csv(output_path    + "customer_data.csv",    index=False)
product_data.to_csv(output_path     + "product_data.csv",     index=False)
df.to_csv(output_path               + "superstore_clean.csv", index=False)

print("\n✅ All files exported successfully.")
print(f"   transaction_data.csv  → SQL project")
print(f"   customer_data.csv     → SQL project")
print(f"   product_data.csv      → SQL project")
print(f"   superstore_clean.csv  → Power BI dashboard")
```
